Q16. Student Performance Analysis

```python
#  A. DATA CLEANING
#Identify missing values
import numpy as np
import pandas as pd
student = pd.read_csv("/content/drive/MyDrive/DSA/Assessment 2/student_performance.csv")

missing_val = student.isnull().sum()
print(missing_val)

student['Age'] = student['Age'].fillna(student['Age'].mean())
student['Attendance'] = student['Attendance'].fillna(student['Attendance'].mean())
student['Python_Score'] = student['Python_Score'].fillna(student['Python_Score'].mean())
student['SQL_Score'] = student['SQL_Score'].fillna(student['SQL_Score'].mean())
student['Pandas_Score'] = student['Pandas_Score'].fillna(student['Pandas_Score'].mean())
student['NumPy_Score'] = student['NumPy_Score'].fillna(student['NumPy_Score'].mean())
student['Final_Score'] = student['Final_Score'].fillna(student['Final_Score'].mean())

missing_val = student.isnull().sum()
print(missing_val)

# B. Feature Engineering

student['Average_Score'] = (student['Python_Score']+student['SQL_Score']+student['Pandas_Score']+student['NumPy_Score'])/4
student['Pass_Status'] = np.where(student['Average_Score']>=50,'Pass','Fail')

# C. Analysis

print("Top 5 students:",student.sort_values(by='Average_Score',ascending=False).head())
print('Department with highest avg score:',student.groupby('Department')['Average_Score'].head(1))
print("Average attendance by dept",student.groupby('Department')['Attendance'].mean())
print("No. of students who passed and failed",student.groupby('Pass_Status')['Student_ID'].count())


# D. Summary

print("Summary: \n",student.describe())
```
